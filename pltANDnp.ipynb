```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel


def model(x):
    return x * np.sin(x)


# Training data
xdata = np.array([1, 3, 5, 6, 8])
ydata = model(xdata)


# Kernel
kernel = ConstantKernel(1.0) * RBF(
    length_scale=1.0,
    length_scale_bounds=(1e-2, 10)
)


# Gaussian Process model
gp = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=10
)


# Fit model
gp.fit(xdata[:, np.newaxis], ydata)


# Prediction points
xfit = np.linspace(0, 10, 1000)


# Predict mean and uncertainty
yfit, std = gp.predict(
    xfit[:, np.newaxis],
    return_std=True
)


# 95% confidence interval
dyfit = 2 * std


# Plot
plt.figure(figsize=(8, 5))

# Confidence interval
plt.fill_between(
    xfit,
    yfit - dyfit,
    yfit + dyfit,
    color='gray',
    alpha=0.25
)

# Mean prediction curve
plt.plot(
    xfit,
    yfit,
    color='dimgray',
    linewidth=1.5
)

# Data points
plt.plot(
    xdata,
    ydata,
    'o',
    color='red',
    markersize=6
)


# Axes formatting
plt.xlim(0, 10)
plt.ylim(-10, 35)

plt.grid(True, linestyle='--', alpha=0.4)

plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

np.partition takes an array and a number K; the result is a new array with the smallest K values to the left of the partition, and the remaining values to the right, in arbitrary order:

```python
x = np.array([7, 2, 3, 1, 6, 5, 4])
np.partition(x, 3)
```

```python
# Use a compound data type for structured arrays
data = np.zeros(4, dtype={'names':('name', 'age', 'weight'),
                          'formats':('U10', 'i4', 'f8')})
print(data.dtype)
```

```python
name = ['Alice', 'Bob', 'Cathy', 'Doug']
age = [25, 45, 37, 19]
weight = [60.0, 85.5, 68.0, 61.5]
```

```python
data['name'] = name
data['age'] = age
data['weight'] = weight
print(data)
```

```python
data[data['age'] < 30]['name']
```
